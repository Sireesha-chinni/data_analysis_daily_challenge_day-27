# Coding Challenge: Ride Booking Data Analysis (Using Tuples)

Scenario


You are working as a data analyst for a ride-booking company (like Ola/Uber).
You received trip data stored as tuples.


Each tuple contains:
(Trip_ID, City, Distance_km, Fare_amount)


Sample Data
trips = (
(101, "Bangalore", 12.5, 250),
(102, "Hyderabad", 8.0, 180),
(103, "Bangalore", 15.2, 300),
(104, "Chennai", 5.5, 120),
(105, "Bangalore", 7.8, 200)
)

Task 1: Total Trips
    
Find the total number of trips.

```python
trips = (
(101, "Bangalore", 12.5, 250),
(102, "Hyderabad", 8.0, 180),
(103, "Bangalore", 15.2, 300),
(104, "Chennai", 5.5, 120),
(105, "Bangalore", 7.8, 200)
)
```

```python
print("Total number of Trips:",len(trips))
```

Task 2: Trips from Bangalore
    
    
Count how many trips happened in "Bangalore".

```python
count =0

for i in trips:
    if "Bangalore" in i:
        count =count+1
        
print("Trips Happened in Bangalore:",count)
```

Task 3: Total Revenue
    

Calculate the total fare amount collected from all trips.

```python
tot_fre = 0

for i in trips:
    tot_fre = tot_fre + i[3]
    
print("Total fare amount collected from all trips:",tot_fre)
```

Task 4: High Distance Trips
    
    
Print all trips where distance > 10 km.

```python
for i in trips:
    if i[2] > 10:
        print(i)
```

Task 5: Highest Fare Trip
    
    
Find the trip with the maximum fare.

```python
fare = []

for i in trips:
    fare.append(i[3])
    
print("The Trip with maximum fare:",trips[fare.index(max(fare))])
```

Task 6 (Bonus - Real Analyst Thinking
        
        
Find the average fare for Bangalore trips only.

```python
avg_list = []
count = 0

for i in trips:
    if "Bangalore" in i:
        count =count+1
        avg_list.append(i[3])
        
print("The Average fare for Bangalore Trips:",sum(avg_list)/count)
```
